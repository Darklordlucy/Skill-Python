## Format specifiers

```python
num_1 = 12345.6789
num_2 = 123.4567
num_3 = 12.3456

print(f"The number is shown as ${num_1:.2f}")     # Round of you that many spaces 
print(f"The number is shown as ${num_2:.2f}")
print(f"The number is shown as ${num_3:.2f}")


print(f"The number is shown as ${num_1:13}")      # Allotate that may spaces  
print(f"The number is shown as ${num_2:13}")
print(f"The number is shown as ${num_3:13}")

print(f"The number is shown as ${num_1:012}")     # Zero padding before the var
print(f"The number is shown as ${num_2:012}")
print(f"The number is shown as ${num_3:012}")

print(f"The number is shown as ${num_1:<4}")     # left adjustment
print(f"The number is shown as ${num_2:<4}")
print(f"The number is shown as ${num_3:<4}")


print(f"The number is shown as ${num_1:>10}")     # Right adjustment
print(f"The number is shown as ${num_2:>10}")
print(f"The number is shown as ${num_3:>10}")

print(f"The number is shown as ${num_1:^10}")     # Centre adjustment
print(f"The number is shown as ${num_2:^10}")
print(f"The number is shown as ${num_3:^10}")

print(f"The number is shown as ${num_1:+}")       # Add + sign at the beginning
print(f"The number is shown as ${num_2:+}")
print(f"The number is shown as ${num_3:+}")

print(f"The number is shown as ${num_1:=}")       # place sign to the left most
print(f"The number is shown as ${num_2:=}")
print(f"The number is shown as ${num_3:=}")
```

#### Arbitary Arguments 

```python
# *args     = allow you to pass multiple non-key arguments ( arg = treat like a list )
# **kwargs  = allow you pass muliple keywords arguments    ( kwargs = treat like a dictionary )

def sum(*args):
    total = 0
    for num in args:
        total += num
    return total

print(sum(1, 2, 3))

def sum_kwargs(**kwargs):
    total = 0
    for value in kwargs.values():
        total += value
    return total

print(sum_kwargs(a=1, b=2, c=7))


    
```

#### Super Function

" The above is used to call parent functions in the child function "

```python
class shape:
    def __init__(self,color,filled):
        self.color = color
        self.filled = filled

class circle(shape):
    def __init__(self, color, filled,radius):
        super().__init__(color, filled)                          # ----> Super function
        self.radius = radius
```

#### Polymorphism

```python
class shape:
    def area(self):
        pass

class circle(shape):
    def __init__(self,radius):
        self.radius = radius 
    def area(self):
        return 3.14*self.radius**2

class square(shape):
    def __init__(self,width):
        self.width = width
    def area(self):
        return self.width*self.width
    
shapes = [circle(3),square(4)]
print(shapes[0])
        
```

#### Static Method 

" It belong to the class but not to an object "

```python
class employee:
    def __init__(self,position):
        self.postion = position
    def get_info(self):
        return f'{self.postion}'
    @staticmethod
    def is_valid_position(position):
        valid_position = ["Cook","Pirate","Swordsman"]
        return position in valid_position

employee1 = employee("Cook")
print(employee.is_valid_position(employee1))
employee1.get_info()
```

#### Property

```python
class rectangle:
    def __init__(self,width,height):
        self._width = width
        self._height = height
    
    @property
    def width(self):
        return f"{self._width : .1f}cm "
    @property
    def height(self):
        return f"{self._height : .1f}cm "
    @height.setter
    def height(self,new_height):
        self.new_height = new_height 
    @width.deleter
    def width(self):
        del self._width
        print("Widht had been deleted")
    
rectangle1 = rectangle(4.33, 3.34)
print(rectangle1.width)
```

#### Decorator

"A function that extends the behavior of another function without modifying the base function"

```python
def add_sprinkles(function):
    def wrapper(*args,**kwargs):
        print("You add sprinkle !")
        function(*args,**kwargs)
    return wrapper

def add_fudge(function):
    def wrapper(*args,**kwargs):
        print("You add fudge !")
        function(*args,**kwargs)
    return wrapper

@add_sprinkles         # ----> Decorator
@add_fudge             # ----> Decorator

def get_ice_icream(flavor):
    print(f"Here is your {flavor}")
get_ice_icream("Vanilla")
```
